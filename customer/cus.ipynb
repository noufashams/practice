```python
# Q6. Create a bar chart showing the number of individuals 
# in the top 10 most common professions (prof).
result=df["prof"].value_counts().head(10)
plt.bar(result.index,result.values)
plt.xlabel("profession")
plt.ylabel("individuals")
plt.title("top 10 prof-counts")
plt.xticks(rotation=90)
plt.show()
```

```python
# Q7. Create a bar chart showing the number of individuals 
# in each location and rotate the x-axis labels.
result=df["loc"].value_counts()
plt.bar(result.index,result.values,color="green")
plt.xlabel("location")
plt.ylabel("individuals")
plt.title("population distribution ")
plt.xticks(rotation=45)
plt.show()

```

```python
# Q8. Using groupby(), calculate the average salary by location 
# and visualize the result with a bar chart.
result=df.groupby("loc")["salary"].mean()
plt.bar(result.index,result.values)
plt.xlabel("location")
plt.ylabel("avg-salary")
plt.title("loc-avg-salary")
plt.show()
```

```python
# Q9. Using groupby(), calculate the average age by location and visualize it with a bar chart.
result=df.groupby("loc")["age"].mean()
plt.bar(result.index,result.values,color="orange")
plt.xlabel("location")
plt.ylabel("avg-age")
plt.title("loc-avg-age")
plt.show()
```

```python
# Q10. Using groupby(), calculate the average salary for each location and customize 
# the graph title, labels, figure size, and grid.
result=df.groupby("loc")["salary"].mean()
plt.figure(figsize=(10,10))
plt.bar(result.index,result.values,color="orange")
plt.xlabel("location")
plt.ylabel("avg-salary")
plt.title("loc-avg-salary")
plt.grid()
plt.show()
```

```python
# Q11. Create a bar chart of the average salary by location, sorted from highest to lowest.
result=df.groupby("loc")["salary"].mean().sort_values(ascending=False)
plt.bar(result.index,result.values,color="pink")
plt.xlabel("location")
plt.ylabel("avg-salary")
plt.title("loc-avg-salary")
plt.show()
```

```python
# Q12. Create a pie chart showing the proportion of individuals in each location.
result=df["loc"].value_counts()
plt.figure(figsize=(8,5))
plt.pie(result.values,labels=result.index,autopct="%1.1f%%")
plt.title("Proportion of Individuals by Location")
plt.show()
```

```python
# Q13. Create a line graph comparing the average salary across locations after calculating the averages with groupby().
result=df.groupby("loc")["salary"].mean()
plt.plot(result.index, result.values)
plt.xlabel("Location")
plt.ylabel("salary")
plt.title("Aerage salary by Location")
plt.show()
```

```python
# Q14. Create a bar chart comparing average age and average salary scaled appropriately for each location.
result=df.groupby("loc")[["salary","age"]].mean()
result.plot(kind="bar", secondary_y=["age"],figsize=(15,15),rot=0)

plt.xlabel("Location")

plt.ylabel("Average Value")

plt.title("Average Age and Salary by Location")

plt.show()
```

```python
# Q15. Create a histogram of salary and customize its figure size, number of bins, transparency, and grid.
plt.figure(figsize=(10,10))
plt.hist(df["salary"],bins=10,alpha=0.5)
plt.grid()
plt.show()
```

```python
# Q16. Create a bar chart of location-wise counts and customize the title, axis labels, tick rotation, figure size, and grid.
result = df["loc"].value_counts()

plt.figure(figsize=(10, 6))
plt.bar(result.index, result.values)
plt.title("Location-wise Count")
plt.xlabel("Location")
plt.ylabel("Number of Individuals")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
# Q17. Create a graph for the average salary by location and add a 
# suitable legend, annotations, and customized tick labels.
result=df.groupby("loc")["salary"].mean()
plt.figure(figsize=(10,10))
plt.pie(result.values,autopct="%1.1f%%")
plt.legend(result.index)
plt.show()
```

```python
result = df.groupby("loc")["salary"].mean()
plt.barh(result.index, result.values)
plt.xlabel("Average Salary")
plt.ylabel("Location")
plt.title("Average Salary by Location")
plt.show()
```
