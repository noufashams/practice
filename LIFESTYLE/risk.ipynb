```python
X=df.drop("health_risk",axis=1)
y=df["health_risk"]
```

```python
X.dtypes
```

```python
# Encoding >>> converting the string values in X into numerical values
# the column profession has multiple values so we are choosing one-hot encoding
# one-hot encoding takes each profession from profession column and coverts to 0 or 1 
# for all the rows by adding new columns for each profession

X=pd.get_dummies(X)
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
# weight>>
# age>>
# height>>

# Feature scaling
# ==================
# feature scaling in machine learning is the process of changing numerical
# columns to use the same value range
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
# scaler.fit(X_train)mean+std
# scaler.transform(X_train)
# OR TOGETHER
X_train=scaler.fit_transform(X_train)#need to fit before transform for X_train
X_test=scaler.transform(X_test)# X_test should not be given in fit()
# if we allow to fit X_test it may lead to data leakage
# fit():Learns and stores stats like the mean and standard deviation from your dataset,
# but does not change the data
# transfrom():Uses the pre calculated stats to scale or change the data
```

```python
from sklearn.neighbors import KNeighborsClassifier
model=KNeighborsClassifier(n_neighbors=5)
model.fit(X_train,y_train)
y_pred=model.predict(X_test)

```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pred))
```
