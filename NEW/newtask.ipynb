```python
# Level 1 – Filtering & Selection
# 1. Display the employees whose salary is greater than 60,000.
# 2. Find all employees who have more than 5 years of experience.
# 3. Display employees working in the IT department with a performance score greater than 8.
# 4. Find employees whose attendance percentage is below 85.
# 5. Display employees who have completed more than 10 projects.
# 6. Find employees working in either Kochi or Bangalore.
# 7. Display employees whose satisfaction score is 8 or above and whose experience is at least 5 years.
# 8. Find employees whose salary is between 40,000 and 70,000.
```

```python
# 1. Display the employees whose salary is greater than 60,000.
e_df[e_df["salary"]>60000][["first_name","salary"]]
```

```python
# 2. Find all employees who have more than 5 years of experience.
e_df[e_df["experience_years"]>5][["first_name","experience_years"]]
```

```python
# 3. Display employees working in the IT department with a performance score greater than 8.
e_df[(e_df["department"]=="IT") & (e_df["satisfaction_score"]>8)]
```

```python
# 4. Find employees whose attendance percentage is below 85.
e_df[e_df["attendance_percentage"]<85]
```

```python
# 5. Display employees who have completed more than 10 projects.
e_df[e_df["projects_completed"]>10]
```

```python
# 6. Find employees working in either Kochi or Bangalore.
e_df[(e_df["city"]=="Kochi") | (e_df["city"]=="Banglore")][["first_name","city"]]
```
