```python
import pandas as pd
data=pd.read_csv("weather_classification_data.csv")
df=pd.DataFrame(data)
df
```

```python
df.info()
```

```python
X=df.drop("Weather Type",axis=1)
y=df["Weather Type"]
X
```

```python
X.dtypes
```

```python
X=pd.get_dummies(X)
X
```

```python
# Encoding target column using label encoding(Nave bayes)
# =======================================================
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
y=le.fit_transform(y)
# print(le.classes_)
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
# KNN>>Scaling is required
# Naive Bayes sclaing is usually not required
```

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
y_pred
```

```python
le.inverse_transform(y_pred)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pred))

# accuracy score calculation---->number of correct predictions/total no. of elements in y_test
# returns the accuracy score
```
