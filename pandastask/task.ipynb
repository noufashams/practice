```python
# Part E – GroupBy and Aggregation
# 35. Find the average salary for each profession using groupby().
# 36. Find the maximum salary for each location using groupby().
# 37. Find the minimum salary for each profession using groupby().
# 38. Find the number of employees in each profession using groupby().
# 39. Find the average salary of employees in each location and sort the result from highest to lowest.
# 40. Find the average salary and average age for each profession using groupby() and agg().
```

```python
# 35. Find the average salary for each profession using groupby().
df.groupby("prof")["salary"].mean()
```

```python
# 36. Find the maximum salary for each location using groupby().
df.groupby("loc")["salary"].max()
```

```python
# 37. Find the minimum salary for each profession using groupby().
df.groupby("prof")["salary"].min()
```

```python
# 38. Find the number of employees in each profession using groupby().
df.groupby("prof")["id"].count()
```

```python
# 39. Find the average salary of employees in each location and sort the result from highest to lowest.
df.groupby("loc")["salary"].mean().sort_values(ascending=False)
```

```python
# 40. Find the average salary and average age for each profession using groupby() and agg().
df.groupby("prof")[["salary","age"]].agg("mean")
```
