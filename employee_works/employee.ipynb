```python
#group by
# ==========
# splits a dataframe into groups based on one or 
# more columns,applies aggregate functions(like a sum or average)
# and combines the results back into a summary DataFrame

#retrn avg age in each location
df.groupby("location")["age"].mean()
```

```python
#return maximum age in each job


df.groupby("job")["age"].max()
```

```python
# aggregate functions
# ======================
# sum(),count(),min(),max(),mean(),size()
# .count() >> returns the non missing values from df given
# .size() >> returns the total rows from df column given
```

```python
#count of jobs
df.groupby("job")["job"].count()#do not count rows with missing values
df.groupby("job")["job"].size()#counts even if missing values are in rows
df.groupby("job")["age"].agg(["mean","min","max","size"])#agg returns multiple aggregste function results
```

```python
df.groupby(["job","location"]).size()
# job           location 
# Accountant    africa       2
#               india        5
# Actor         africa       1
#               india        2
#               ireland      1
#                           ..
# Veterinarian  us           1
# Writer        australia    1
#               china        1
#               india        3
# africa        india        1
# Length: 198, dtype: int64

df.groupby("job")["age"].mean().sort_values(ascending=False)

df.groupby(["job","location"])["age"].mean().sort_values(ascending=False)

#avg and max of age in a job
df.groupby("job")["age"].agg(["mean","max"])
```

```python
minimum=df["age"].min()
maximum=df["age"].max()
average=df["age"].mean()

labels=["minimum","maximum","average"]
values=[minimum,maximum,average]
plt.bar(labels,values,width=0.5)#(x,y)vertical bar chart width can be changed
plt.title("age")
plt.show()
```

```python
minimum=df["age"].min()
maximum=df["age"].max()
average=df["age"].mean()

labels=["minimum","maximum","average"]
values=[minimum,maximum,average]
plt.barh(labels,values,color="pink",height=0.5)#horizontal bar chart(y,x)height can be changed
plt.title("age")
plt.show()
```
